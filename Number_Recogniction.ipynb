```python
# import libraries
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
from tensorflow import keras
```

```python
# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

# Reshape the data to match the expected dimensions
x_train = x_train.reshape(60000, 28 * 28)
x_test = x_test.reshape(10000, 28 * 28)

# Assertions to verify the shapes
assert x_train.shape == (60000, 28 * 28)
assert x_test.shape == (10000, 28 * 28)
assert y_train.shape == (60000,)
assert y_test.shape == (10000,)

# Print the shapes
print(x_train.shape, y_train.shape)
print(x_test.shape, y_test.shape)
```

```python
X = x_train/255
y = y_train

print(X.shape)
print(y.shape)
```

```python
# Convert array to tensor
X = torch.tensor(X, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.long)
```

```python
# Build ANN Model
def model(X, w1, b1, w2, b2, w3, b3, w4 , b4 , w5 , b5 , w6 , b6):
  # layer 1
  z1 = torch.matmul(X, w1) + b1
  a1 = torch.sigmoid(z1)

  # layer 2
  z2 = torch.matmul(a1, w2) + b2
  a2 = torch.sigmoid(z2)

  # layer 3
  z3 = torch.matmul(a2, w3) + b3
  a3 = torch.sigmoid(z3)

  # layer 4
  z4 = torch.matmul(a3, w4) + b4
  a4 = torch.sigmoid(z4)

  # layer 5
  z5 = torch.matmul(a4, w5) + b5
  a5 = torch.sigmoid(z5)

  # layer 6
  z6 = torch.matmul(a5, w6) + b6

  return z6
```

```python
# Training
# Initial weights
w1 = torch.randn((784, 256), dtype=torch.float32, requires_grad=True)
b1 = torch.randn((1, 256), dtype=torch.float32, requires_grad=True)
w2 = torch.randn((256, 128), dtype=torch.float32, requires_grad=True)
b2 = torch.randn((1, 128), dtype=torch.float32, requires_grad=True)
w3 = torch.randn((128, 64), dtype=torch.float32, requires_grad=True)
b3 = torch.randn((1, 64), dtype=torch.float32, requires_grad=True)
w4 = torch.randn((64, 32), dtype=torch.float32, requires_grad=True)
b4 = torch.randn((1, 32), dtype=torch.float32, requires_grad=True)
w5 = torch.randn((32, 16), dtype=torch.float32, requires_grad=True)
b5 = torch.randn((1, 16), dtype=torch.float32, requires_grad=True)
w6 = torch.randn((16, 10), dtype=torch.float32, requires_grad=True)
b6 = torch.randn((1, 10), dtype=torch.float32, requires_grad=True)

# Optimizer
optimizer = optim.Adam([w1, b1, w2, b2, w3, b3 ,w4 ,b4 ,w5 ,b5 ,w6 ,b6], lr=0.005)

# Cost function
cost_fn = nn.CrossEntropyLoss()

# Training loop
for epoch in range(400):
  batch_size = 1200
  for i in range(0, len(X), batch_size):
    X_batch = X[i:i+batch_size]
    y_batch = y[i:i+batch_size]
    y_pred = model(X_batch, w1, b1, w2, b2, w3, b3 ,w4 ,b4 ,w5 ,b5 ,w6 ,b6)
    loss = cost_fn(y_pred, y_batch)
    loss.backward()
    optimizer.step()
    optimizer.zero_grad()
    print(f'Epoch: {epoch+1:03d}, Batch number: {i:03d}, Loss: {loss.item():.4f}')
```

```python
X_test = x_test/255
y_test = y_test
print(X_test.shape)
print(y_test.shape)

tX_test = torch.tensor(X_test, dtype=torch.float32)
ty_test = torch.tensor(y_test, dtype=torch.long)
```

```python
# Evaluate the model using the test set
with torch.no_grad():
  h = model(tX_test, w1, b1, w2, b2, w3, b3 ,w4 ,b4 ,w5 ,b5 ,w6 ,b6)
  y_pred = torch.argmax(h, dim=1)
  accuracy = (y_pred == ty_test).sum().item() / len(ty_test)
  print(f'Model accuracy = {accuracy*100:.2f}%')
```
